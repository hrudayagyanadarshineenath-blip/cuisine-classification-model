```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

print("Libraries imported successfully.")
```

```python
file_path = r"C:\Users\hruda\anaconda_projects\d8f64f1b-06e2-493e-93e3-57826b3b231f\Dataset -1.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
display(df.head())
```

```python
print("Dataset columns:")
print(df.columns.tolist())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Number of unique cuisine values:", df["Cuisines"].nunique())

print("\nTop 20 cuisine categories:")
print(df["Cuisines"].value_counts().head(20))
```

```python
classification_df = df.dropna(subset=["Cuisines"]).copy()

print("Dataset after removing missing cuisine values:")
print("Shape:", classification_df.shape)
```

```python
cuisine_counts = classification_df["Cuisines"].value_counts()

print("Number of cuisine classes:", len(cuisine_counts))

display(cuisine_counts.head(20))
```

```python
features = [
    "Country Code",
    "City",
    "Locality",
    "Longitude",
    "Latitude",
    "Average Cost for two",
    "Currency",
    "Has Table booking",
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu",
    "Price range",
    "Aggregate rating",
    "Votes"
]

X = classification_df[features]
y = classification_df["Cuisines"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Input features:")
display(X.head())

print("\nTarget cuisine:")
display(y.head())
```

```python
numerical_features = [
    "Country Code",
    "Longitude",
    "Latitude",
    "Average Cost for two",
    "Price range",
    "Aggregate rating",
    "Votes"
]

categorical_features = [
    "City",
    "Locality",
    "Currency",
    "Has Table booking",
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu"
]

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

print("Preprocessing transformers created successfully.")
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
classifier = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Random Forest classifier created successfully.")
```

```python
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", classifier)
    ]
)

print("Classification pipeline created successfully.")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully.")
```

```python
y_pred = model.predict(X_test)

print("Predictions completed successfully.")
print("Number of predictions:", len(y_pred))
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 4))
```

```python
precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
```

```python
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
performance = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall"
    ],
    "Score": [
        accuracy,
        precision,
        recall
    ]
})

display(performance)
```

```python
comparison = pd.DataFrame({
    "Actual Cuisine": y_test.values,
    "Predicted Cuisine": y_pred
})

display(comparison.head(20))
```

```python
correct_predictions = (y_test.values == y_pred).sum()
total_predictions = len(y_test)

print("Correct predictions:", correct_predictions)
print("Total predictions:", total_predictions)
print(
    "Correct prediction percentage:",
    round((correct_predictions / total_predictions) * 100, 2),
    "%"
)
```

```python
comparison.to_csv(
    "cuisine_classification_predictions.csv",
    index=False
)

print("Predictions saved successfully.")
```
